## Exercise 2: Safe casting challenge

- Given `values = ["25", "3.14", "-8", "hello"]`, decide how each text should be converted.
- Use string methods (such as `isdigit()`, `replace()`, `lstrip()`) and `if/elif/else` to check each value.
- Convert to `int` if it is a whole number, to `float` if it is a decimal number, otherwise keep it as a string.
- Print each converted value together with its type.

```python
# Write your solution here

values =["25","3.14","-8","hello"]

for value in values:
    if value.lstrip("-").isdigit():
        converted =int(value)
    elif value.lstrip("-").replace(".","",1).isdigit():
        converted =float(value)
    else:
        converted=value

    print(converted, type(converted))
```

## Exercise 4: String transformation pipeline

- Start with `raw = "   python,java,c++ , go  "`.
- Strip the outer spaces and split the text by comma.
- Print the split list.
- Loop through the list, strip each language, uppercase it, and print it.

```python
# Write your solution here
raw ="python , java, c++, go"

languages = raw.strip().split(",")
print(languages)

for language in languages:
    cleaned_language = language.strip().upper()
    print(cleaned_language)
```
