## Exercise 4: Comprehension and sort

1. From `scores = [75, 88, 92, 60, 81]`, create `high_scores` with only scores >= 80 using list comprehension.
2. Sort `high_scores` in ascending order.
3. Sort it again in descending order.
4. Print all results.

```python
# Write your solution here
print("Loop through list")
scores = [75, 88, 92, 60, 81]
for score in scores:
    print(score)

print ("Scores greater than or equal to 80:")    
for score in scores:
    if score >= 80:
        print(score)
        
total_score = sum(scores)
print ("Total score:", total_score)
```

## Exercise 5: Copy and join lists

1. Create `list_a = [1, 2, 3]` and `list_b = [4, 5, 6]`.
2. Make a copy of `list_a` in a new variable.
3. Join `list_a` and `list_b` into one list in two ways:
   - using `+`
   - using `extend()`
4. Print all lists.

```python
# Write your solution here
list_a = [1, 2, 3]
list_b = [4, 5, 6]

copied_list = list_a.copy()
print("Copied list:", copied_list)

joined_list = list_a + list_b
print("Joined using +:", joined_list)

extended_list = list_a.copy()
extended_list.extend(list_b)
print("Joined using extend:", extended_list)

print("List A:", list_a)
print("List B:", list_b)
```
