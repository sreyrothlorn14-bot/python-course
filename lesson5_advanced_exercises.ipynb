## Exercise 5: Set from a list comprehension

- Use a list comprehension to build the squares of numbers 1..20 where the square is even.
- Convert the list to a set to remove duplicates (if any).
- Print the resulting set and its length.
- Convert it to a sorted list and print.

```python
# Write your solution here
squares_list = [number ** 2 for number in range(1, 21) if number ** 2 % 2 == 0]

squares_set = set(squares_list)

print("Set:", squares_set)
print("Length:", len(squares_set))

sorted_squares = sorted(squares_set)
print("Sorted list:", sorted_squares)
```
