## Exercise 1: Robust numeric parser

- Write a function that parses user-provided string to int or float.
- Raise/handle exceptions for invalid formats.
- Return tuple `(success, value_or_error)` and print examples.

```python
# Exercise 1
def parse_number(text):
    try:
        if "." not in text:
            return True, int(text)
        return True, float(text)
    except (TypeError, ValueError) as error:
        return False, str(error)

print(parse_number("42"))
print(parse_number("3.14"))
print(parse_number("hello"))
```

## Exercise 2: Multi-exception workflow

- Build a division function that converts both inputs with `float()` and divides, using `try/except/else/finally`.
- Handle `TypeError`, `ValueError`, and `ZeroDivisionError` separately.
- Demonstrate all branches with different calls.

```python
# Exercise 2
def divide(a, b):
    try:
        left = float(a)
        right = float(b)
        result = left / right
    except TypeError:
        print("TypeError: inputs must be convertible to float")
    except ValueError:
        print("ValueError: invalid numeric text")
    except ZeroDivisionError:
        print("ZeroDivisionError: cannot divide by zero")
    else:
        print("Result:", result)
    finally:
        print("Division attempt finished")

divide("10", "2")
divide(None, "2")
divide("abc", "2")
divide("10", "0")
```

## Exercise 3: None-aware processing

- Given list with values and `None`, compute average of valid numbers only.
- Ignore `None` safely.
- Print cleaned list and average.

```python
# Exercise 3
values = [12, None, 18, 7, None, 23, 10]
cleaned = [value for value in values if value is not None]
average = sum(cleaned) / len(cleaned) if cleaned else None

print("Cleaned list:", cleaned)
print("Average:", average)
```

## Exercise 4: Input validation loop

- Prompt repeatedly for integer between 1 and 100 until valid.
- Use exception handling and custom error messages.
- Print accepted value and attempt count.

```python
# Exercise 4
attempt_count = 0

while True:
    try:
        value = int(input("Enter an integer between 1 and 100: "))
        attempt_count += 1

        if not 1 <= value <= 100:
            print("Error: value must be between 1 and 100")
+            continue
        print("Accepted value:", value)
        print("Attempt count:", attempt_count)
        break
    except ValueError:
        attempt_count += 1
        print("Error: please enter a whole number")
```

## Exercise 5: Mini CLI app

- Create text menu app (add/subtract/multiply/divide/quit).
- Use input + try/except for invalid choices and values.
- Keep looping until quit command.

```python
# Exercise 5
while True:
    print("\nMenu: add, subtract, multiply, divide, quit")
    choice = input("Choose an operation: ").strip().lower()

    if choice == "quit":
        print("Goodbye")
        break
    if choice not in ("add", "subtract", "multiply", "divide"):
        print("Invalid choice")
        continue

    try:
        left = float(input("Enter the first number: "))
        right = float(input("Enter the second number: "))

        if choice == "add":
            result = left + right
        elif choice == "subtract":
            result = left - right
        elif choice == "multiply":
            result = left * right
        else:
            if right == 0:
                raise ZeroDivisionError
            result = left / right
    except ZeroDivisionError:
        print("Error: cannot divide by zero")
    except ValueError:
        print("Error: please enter numeric values")
    else:
        print("Result:", result)
```
