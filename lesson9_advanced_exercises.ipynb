## Exercise 1: Flexible calculator

- Write `calculate(a, b, op)` to handle `+`, `-`, `*`, `/`.
- Return result or error message for unsupported operator.
- Demonstrate function with multiple calls.

```python
# Exercise 1
def calculate(a, b, op):
    if op == "+":
        return a + b
    if op == "-":
        return a - b
    if op == "*":
        return a * b
    if op == "/":
        if b == 0:
            return "Error: cannot divide by zero"
        return a / b
    return "Error: unsupported operator"

print(calculate(10, 5, "+"))
print(calculate(10, 5, "-"))
print(calculate(10, 5, "*"))
print(calculate(10, 5, "/"))
print(calculate(10, 5, "%"))
```

## Exercise 2: *args and **kwargs combo

- Write function `report(*args, **kwargs)`.
- Print positional args and key-value kwargs cleanly.
- Call it with at least 3 positional and 3 keyword arguments.

```python
# Exercise 2
def report(*args, **kwargs):
    print("Positional arguments:")
    for index, value in enumerate(args, start=1):
        print(f"  {index}: {value}")

    print("Keyword arguments:")
    for key, value in kwargs.items():
        print(f"  {key}: {value}")

report("Apple", 42, True, name="Sok", age=20, course="Python")
```

## Exercise 3: Higher-order function

- Write `apply_twice(func, value)` that applies function two times.
- Test with one normal function and one lambda.
- Print both results.

```python
# Exercise 3
def apply_twice(func, value):
    return func(func(value))

def add_three(number):
    return number + 3

print(apply_twice(add_three, 5))
print(apply_twice(lambda number: number * 2, 5))
```

## Exercise 4: Scope control

- Create global counter and function that updates it using `global`.
- Create another function using local variable with same name.
- Print outputs to show scope difference.

```python
# Exercise 4
counter = 0

def increment_global():
    global counter
    counter += 1

def local_counter():
    counter = 100
    print("Local counter:", counter)

print("Initial global counter:", counter)
increment_global()
increment_global()
print("Updated global counter:", counter)
local_counter()
print("Global counter after local function:", counter)
```

## Exercise 5: Recursive problem

- Write recursive `fibonacci(n)`.
- Print fibonacci sequence for n = 0..10.
- Add memoization (dict cache) to optimize repeated calls.

```python
# Exercise 5
fibonacci_cache = {0: 0, 1: 1}

def fibonacci(n):
    if n in fibonacci_cache:
        return fibonacci_cache[n]
    fibonacci_cache[n] = fibonacci(n - 1) + fibonacci(n - 2)
    return fibonacci_cache[n]

sequence = [fibonacci(n) for n in range(11)]
print("Fibonacci sequence:", sequence)
```
