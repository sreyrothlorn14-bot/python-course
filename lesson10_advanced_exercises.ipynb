## Exercise 1: Array statistics

- Create numeric array/list with at least 12 elements.
- Compute min, max, sum, average, and median (manual or built-in helper).
- Print all metrics.

```python
# Exercise 1
numbers = [14, 3, 28, 9, 17, 5, 22, 11, 30, 8, 19, 13]

sorted_numbers = sorted(numbers)
length = len(sorted_numbers)
middle = length // 2
median = (sorted_numbers[middle - 1] + sorted_numbers[middle]) / 2 if length % 2 == 0 else sorted_numbers[middle]

print("Min:", min(numbers))
print("Max:", max(numbers))
print("Sum:", sum(numbers))
print("Average:", sum(numbers) / length)
print("Median:", median)
```

## Exercise 2: Iterator protocol deep dive

- Create iterator from a string and consume values with `next()` until exhausted.
- Handle `StopIteration` explicitly.
- Print total number of consumed characters.

```python
# Exercise 2
text = "Python"
text_iterator = iter(text)
count = 0

while True:
    try:
        character = next(text_iterator)
        print(character)
        count += 1
    except StopIteration:
        break

print("Total consumed characters:", count)
```

## Exercise 3: Custom iterator class

- Implement class `EvenNumbers(limit)` yielding even numbers up to limit.
- Use `__iter__` and `__next__`.
- Iterate and print values.

```python
# Exercise 3
class EvenNumbers:
    def __init__(self, limit):
        self.limit = limit
        self.current = 0

    def __iter__(self):
        return self

    def __next__(self):
        if self.current > self.limit:
            raise StopIteration
        value = self.current
        self.current += 2
        return value

for number in EvenNumbers(20):
    print(number)
```

## Exercise 4: Iterator state with `next()`

- Create a custom iterator class `Countdown(start)` that counts down to 1.
- Use `iter()` and `next()` to take the first two values manually.
- Then use a `for` loop on the same iterator to get the remaining values.
- Add a short comment explaining why the `for` loop continues where `next()` stopped.

```python
# Exercise 4
class Countdown:
    def __init__(self, start):
        self.current = start

    def __iter__(self):
        return self

    def __next__(self):
        if self.current < 1:
            raise StopIteration
        value = self.current
        self.current -= 1
        return value

countdown = Countdown(5)
print(next(iter(countdown)))
print(next(countdown))

# The for loop continues at the current state because it uses the same
# iterator object, so the previous next() calls do not reset it.
for number in countdown:
    print(number)
```
