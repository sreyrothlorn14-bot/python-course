## Exercise 1: Array basics (using list)

1. Create `cars = ["Ford", "BMW", "Volvo"]`.
2. Print the first and last elements.
3. Print the number of elements.
4. Loop through and print each car.

```python
# Exercise 1
cars = ["Ford", "BMW", "Volvo"]

print("First:", cars[0])
print("Last:", cars[-1])
print("Number of elements:", len(cars))

for car in cars:
    print(car)
```

## Exercise 2: Array operations

Using `cars`:

1. Append `"Honda"`.
2. Remove `"BMW"`.
3. Pop the last item and print what was removed.
4. Print the updated array.

```python
# Exercise 2
cars.append("Honda")
cars.remove("BMW")
removed = cars.pop()

print("Removed:", removed)
print("Updated array:", cars)
```

## Exercise 3: Built-in iterator

1. Create `nums = [10, 20, 30]`.
2. Create an iterator from `nums` using `iter()`.
3. Use `next()` to print each value one by one.
4. (Optional) Use `try/except` to catch `StopIteration`.

```python
# Exercise 3
nums = [10, 20, 30]
nums_iterator = iter(nums)

while True:
    try:
        print(next(nums_iterator))
    except StopIteration:
        break
```

## Exercise 4: Custom iterator class

1. Create a class `CountToFive`.
2. Implement `__iter__()` and `__next__()`.
3. The iterator should return numbers 1 to 5.
4. Use a `for` loop to print all values.

```python
# Exercise 4
class CountToFive:
    def __init__(self):
        self.current = 1

    def __iter__(self):
        return self

    def __next__(self):
        if self.current > 5:
            raise StopIteration
        value = self.current
        self.current += 1
        return value

for number in CountToFive():
    print(number)
```
